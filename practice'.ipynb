```python
df = pd.DataFrame({
    "quantity":[1,2,3,4,5],
    "price":[80,160,220,340,400]
})
```

```python
df
```

```python
X = df[['quantity']]
X=np.array(X)
```

```python
y = np.array(df['price'])
```

```python
model = LinearRegression()
```

```python
model.fit(X,y)
```

```python
model.predict([[6]])
```

```python
plt.scatter(df["quantity"], df["price"])

plt.xlabel("Quantity")
plt.ylabel("Price")
plt.title("Quantity vs Price")

plt.show()
```

```python
predicted = model.predict(X)

plt.scatter(X, y, label="Actual")
plt.plot(X, predicted, label="Regression Line")

plt.xlabel("Quantity")
plt.ylabel("Price")
plt.title("Linear Regression: Quantity vs Price")
plt.legend()

plt.show()
```
